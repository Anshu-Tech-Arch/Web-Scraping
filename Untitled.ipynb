```python
import requests 
import os 
```

```python
a=requests.get('https://books.toscrape.com/')
```

```python
with open('htmls/page1html', 'w') as f:
    f.write(a.text)
```

```python
for i in range(2,51):
    a=requests.get(f'https://books.toscrape.com/catalogue/page-{i}.html')
    with open(f'htmls/page-{i}.html', 'w', encoding='utf-8') as f:
        f.write(a.text)
```

```python
from bs4 import BeautifulSoup 
```

```python
with open("htmls/page1html.html") as f:
    content=f.read()
```

```python
soup=BeautifulSoup(content, 'html.parser')
```

```python
articles=soup.select("article.product_pod")
```

```python
for article in articles:
    title=article.find('h3').find('a')['title']
    print(title)
```
