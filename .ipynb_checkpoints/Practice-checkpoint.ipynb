```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
books_info=[]
```

```python
for i in range(1,51):
    html=requests.get(f"https://books.toscrape.com/catalogue/page-{i}.html")
    soup=BeautifulSoup(html.text, 'html.parser')
    book=soup.find_all('li', class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")
    for item in book:
        name_tag=item.find("h3").find("a")
        book_name=name_tag.text.strip()
        book_link=name_tag['href']
        price=item.find('p', class_='price_color').text.strip().split('£')[1]
        all_product={
            "Name":book_name,
            "Link":book_link,
            "Price":price
        }
        books_info.append(all_product)
df=pd.DataFrame(books_info)
df.to_csv(f'books_info{i}.csv', index=False)
```
