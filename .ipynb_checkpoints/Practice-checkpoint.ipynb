```python
import requests
from bs4 import BeautifulSoup
```

```python
html=requests.get("https://books.toscrape.com/")
```

```python
with open('htmlsP/page1.html', 'w') as f:
    content=f.write(html.text)
```

```python
books_info=[]
```

```python
soup=BeautifulSoup(html.text, 'html.parser')
```

```python
book=soup.find_all('li', class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")
```

```python
for item in book:
    name_tag=item.find("h3").find("a")
    book_name=name_tag.text.strip()
    book_link=name_tag['href']
    price=item.find('p', class_='price_color').text.strip()
    all_product={
        "Name":book_name,
        "Link":book_link,
        "Price":price
    }
    books_info.append(all_product)
```

```python
import pandas as pd
df=pd.DataFrame(books_info)
```

```python
df.to_csv('books_info.csv', index=False)
```
